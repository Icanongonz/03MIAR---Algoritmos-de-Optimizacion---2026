##Devolución de cambio por técnica voraz

```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Problema: Encontrar los dos puntos más cercanos

```
import random

LISTA_1D = [random.randrange(1,10000) for x in range(1000)]

LISTA_2D = [(random.randrange(1,10000), random.randrange(1,10000))
            for x in range(1000)]

LISTA_3D = [(random.randrange(1,10000), random.randrange(1,10000), random.randrange(1,10000))
            for x in range(1000)]
```

```
def puntos_mas_cercanos_1D_fuerza_bruta(lista):
    min_dist = float('inf')
    mejor_par = None

    for i in range(len(lista)):
        for j in range(i+1, len(lista)):
            dist = abs(lista[i] - lista[j])

            if dist < min_dist:
                min_dist = dist
                mejor_par = (lista[i], lista[j])

    return mejor_par, min_dist


par, distancia = puntos_mas_cercanos_1D_fuerza_bruta(LISTA_1D)

print("Par:", par)
print("Distancia:", distancia)
```

### Complejidad del algoritmo de fuerza bruta en 1D

La estrategia de fuerza bruta consiste en comparar cada elemento de la lista con todos los elementos que aparecen después de él para calcular la distancia entre todos los pares posibles.

Si la lista contiene **n** elementos, el número de comparaciones realizadas es:

$$
\frac{n(n-1)}{2}
$$

Por ejemplo:

- Para n = 10, se realizan 45 comparaciones.
- Para n = 100, se realizan 4.950 comparaciones.
- Para n = 1.000, se realizan 499.500 comparaciones.

Al crecer proporcionalmente a \(n^2\), la complejidad temporal del algoritmo es:

$$
O(n^2)
$$

Esto significa que si duplicamos el número de elementos, el tiempo de ejecución se multiplica exponencialmente.

Por este motivo, aunque la fuerza bruta garantiza encontrar la solución correcta, resulta poco eficiente para conjuntos de datos grandes.

Para mejorar el rendimiento se puede ordenar previamente la lista y comparar únicamente elementos consecutivos, reduciendo la complejidad gracias al uso de la estrategia Divide y Vencerás.

```
def puntos_mas_cercanos_1D(lista):

    ordenada = sorted(lista)

    min_dist = float('inf')
    mejor_par = None

    for i in range(len(ordenada)-1):

        dist = ordenada[i+1] - ordenada[i]

        if dist < min_dist:
            min_dist = dist
            mejor_par = (ordenada[i], ordenada[i+1])

    return mejor_par, min_dist


par, distancia = puntos_mas_cercanos_1D(LISTA_1D)

print("Par:", par)
print("Distancia:", distancia)
```

### Complejidad del algoritmo Divide y Vencerás en 1D

La estrategia utilizada consiste en ordenar previamente la lista y, una vez ordenada, comparar únicamente elementos consecutivos ya que los dos puntos necesariamente aparecerán en posiciones adyacentes de la lista.

El algoritmo realiza dos pasos:

1. Ordenar la lista.
2. Recorrer la lista calculando la distancia entre elementos consecutivos.

El coste de la ordenación es:

$$
O(n \log n)
$$

y el recorrido posterior requiere una única pasada por la lista:

$$
O(n)
$$

Por tanto, la complejidad total es:

$$
T(n)=O(n\log n)+O(n)
$$

Como el término dominante es \(O(n\log n)\), la complejidad final del algoritmo es:

$$
T(n)=O(n\log n)
$$

Comparado con el algoritmo de fuerza bruta, cuya complejidad es:

$$
O(n^2)
$$

la mejora es significativa. Por ejemplo, para una lista de 1000 elementos:

- Fuerza bruta: aproximadamente 500.000 comparaciones.
- Divide y Vencerás: aproximadamente 9.966 operaciones.

Por ello, la solución basada en ordenar la lista resulta mucho más eficiente para conjuntos de datos grandes.

### Posibles mejoras mediante KD-Trees y algoritmos aleatorizados

La solución mediante Divide y Vencerás alcanza una complejidad de:

$$
O(n \log n)
$$

que es óptima para el problema general de los puntos más cercanos. Sin embargo, existen otras técnicas que pueden mejorar el rendimiento práctico en determinados escenarios.

Una de ellas es el uso de **KD-Trees (K-Dimensional Trees)**, estructuras de datos que organizan los puntos del espacio de forma jerárquica. Gracias a esta organización, es posible descartar grandes regiones del espacio durante la búsqueda y reducir significativamente el número de comparaciones necesarias. Aunque la construcción del árbol tiene un coste de:

$$
O(n \log n)
$$

las búsquedas posteriores suelen ser muy eficientes, especialmente en conjuntos de datos grandes.

Otra alternativa son los **algoritmos aleatorizados**, que utilizan técnicas probabilísticas para reducir el número de comparaciones. Estos algoritmos pueden alcanzar, en distribuciones favorables de los datos, una complejidad esperada cercana a:

$$
O(n)
$$

No obstante, al depender del azar, no garantizan dicho comportamiento en todos los casos.

Por ello, aunque KD-Trees y algoritmos aleatorizados pueden ofrecer mejores tiempos de ejecución en determinadas situaciones, la estrategia de Divide y Vencerás sigue siendo la solución clásica más robusta y con mejores garantías teóricas para resolver el problema de los dos puntos más cercanos.

```
import math


# =========================
# DIVIDE Y VENCERÁS EN 2D
# =========================

def distancia_2D(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2
    )


def fuerza_bruta_2D(puntos):
    min_dist = float("inf")
    mejor_par = None

    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            dist = distancia_2D(puntos[i], puntos[j])

            if dist < min_dist:
                min_dist = dist
                mejor_par = (puntos[i], puntos[j])

    return mejor_par, min_dist


def divide_y_venceras_2D(puntos):
    if len(puntos) <= 3:
        return fuerza_bruta_2D(puntos)

    mitad = len(puntos) // 2
    punto_medio = puntos[mitad]

    izquierda = puntos[:mitad]
    derecha = puntos[mitad:]

    par_izquierda, dist_izquierda = divide_y_venceras_2D(izquierda)
    par_derecha, dist_derecha = divide_y_venceras_2D(derecha)

    if dist_izquierda <= dist_derecha:
        mejor_par = par_izquierda
        min_dist = dist_izquierda
    else:
        mejor_par = par_derecha
        min_dist = dist_derecha

    franja = [
        punto for punto in puntos
        if abs(punto[0] - punto_medio[0]) < min_dist
    ]

    franja.sort(key=lambda punto: punto[1])

    for i in range(len(franja)):
        j = i + 1

        while (
            j < len(franja)
            and franja[j][1] - franja[i][1] < min_dist
        ):
            dist = distancia_2D(franja[i], franja[j])

            if dist < min_dist:
                min_dist = dist
                mejor_par = (franja[i], franja[j])

            j += 1

    return mejor_par, min_dist


def puntos_mas_cercanos_2D(lista):
    puntos_ordenados = sorted(lista, key=lambda punto: punto[0])
    return divide_y_venceras_2D(puntos_ordenados)


par_2D, distancia_2D_resultado = puntos_mas_cercanos_2D(LISTA_2D)
print("Par 2D:", par_2D)
print("Distancia 2D:", distancia_2D_resultado)
```

```
# =========================
# DIVIDE Y VENCERÁS EN 3D
# =========================

def distancia_3D(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2 +
        (p1[2] - p2[2]) ** 2
    )


def fuerza_bruta_3D(puntos):
    min_dist = float("inf")
    mejor_par = None

    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            dist = distancia_3D(puntos[i], puntos[j])

            if dist < min_dist:
                min_dist = dist
                mejor_par = (puntos[i], puntos[j])

    return mejor_par, min_dist


def divide_y_venceras_3D(puntos):
    if len(puntos) <= 3:
        return fuerza_bruta_3D(puntos)

    mitad = len(puntos) // 2
    punto_medio = puntos[mitad]

    izquierda = puntos[:mitad]
    derecha = puntos[mitad:]

    par_izquierda, dist_izquierda = divide_y_venceras_3D(izquierda)
    par_derecha, dist_derecha = divide_y_venceras_3D(derecha)

    if dist_izquierda <= dist_derecha:
        mejor_par = par_izquierda
        min_dist = dist_izquierda
    else:
        mejor_par = par_derecha
        min_dist = dist_derecha

    franja = [
        punto for punto in puntos
        if abs(punto[0] - punto_medio[0]) < min_dist
    ]

    franja.sort(key=lambda punto: punto[1])

    for i in range(len(franja)):
        j = i + 1

        while (
            j < len(franja)
            and franja[j][1] - franja[i][1] < min_dist
        ):
            if abs(franja[j][2] - franja[i][2]) < min_dist:
                dist = distancia_3D(franja[i], franja[j])

                if dist < min_dist:
                    min_dist = dist
                    mejor_par = (franja[i], franja[j])

            j += 1

    return mejor_par, min_dist


def puntos_mas_cercanos_3D(lista):
    puntos_ordenados = sorted(lista, key=lambda punto: punto[0])
    return divide_y_venceras_3D(puntos_ordenados)

par_3D, distancia_3D_resultado = puntos_mas_cercanos_3D(LISTA_3D)
print("Par 3D:", par_3D)
print("Distancia 3D:", distancia_3D_resultado)
```
